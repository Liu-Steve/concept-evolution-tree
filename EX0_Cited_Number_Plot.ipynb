# Cited Number Plot

Draw the picture of the citation/article counts needed in the paper.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors


def cited_number_plot(paper_name, citations):
    # Get a list of all years
    years_present = {int(e["key"]): int(e["count"]) for e in citations["group_by"]}
    min_year = min(years_present.keys())
    max_year = max(years_present.keys())

    # Generate a complete year range
    all_years = list(range(min_year // 10 * 10, max_year + 1))

    # Get the number of citations for each year, and fill missing years with 0
    citation_counts = [years_present.get(year, 0) for year in all_years]

    # Set the size of the chart
    plt.figure(figsize=(16, 8))

    # Normalize citation counts to map to colors
    norm = mcolors.PowerNorm(1.2, vmin=min(citation_counts), vmax=max(citation_counts))
    cmap = plt.get_cmap("coolwarm")
    # Calculate the color of each column based on the number of citations
    colors = cmap(norm(citation_counts))

    # Draw a bar chart
    plt.bar(all_years, citation_counts, color=colors)

    # Add a title and tags
    plt.title(paper_name, fontsize=24)
    plt.xlabel("year", fontsize=24)
    plt.ylabel("number", fontsize=24)

    # Set the x-axis scale to each year
    if len(all_years) > 8:
        plt.xticks(
            range(min(all_years), max(all_years) + 1, 10), rotation=0, fontsize=24
        )
    else:
        plt.xticks(all_years, rotation=0, fontsize=24)
    y_max = max(citation_counts)
    if y_max < 100:
        plt.yticks(np.arange(y_max, step=y_max // 8), fontsize=24)
    else:
        plt.yticks(fontsize=24)

    # Add grid lines (y-axis only)
    plt.grid(axis="y", linestyle="--", alpha=0.7)

    # Show Chart
    plt.tight_layout()
    # plt.show()
    plt.savefig(f'./pic_for_paper/{paper_name}.pdf')
```
