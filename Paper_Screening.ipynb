# Paper screening

To ensure that CET's description of the evolution is accurate enough, we need to select a number of high quality papers in OpenAlex. Here we provide two methods for selecting from the recommended conference/journal list.

One approach is to use "primary_location - source - id" from the JSON to determine if the paper is in the list. In this case, you need to find the conference or journal's ID in OpenAlex, which exists as a URL and whose last element starts with an 'S'.

Another approach is selecting it by "booktitle" in the JSON. In this case, you should provide a name string. We recommend trying to search in OpenAlex before formally filtering to determine exactly how their names are stored.

Unfortunately, for a large database like OpenAlex, it is very unlikely that no data will be missed. Based on our experience, in most cases, the first method is more suitable for selecting journal articles, while the second is more suitable for conference papers. We provide implementations of both methods in this notebook.

```python
"""
Determined by "primary_location - source - id"
List of Category A Journals and Conferences Recommended by the China Computer Federation (CCF)
Please create output folder openalex_data at first.
"""

import json

# ID list
valid_ids = [
    "https://openalex.org/S196139623",  # AI
    "https://openalex.org/S199944782",  # TPAMI
    "https://openalex.org/S25538012",  # IJCV
    "https://openalex.org/S118988714",  # JMLR
    "https://openalex.org/S4210191458",  # AAAI
    "https://openalex.org/S4306420609",  # NeurIPS
    "https://openalex.org/S4306420508",  # ACL
    "https://openalex.org/S4363607701",  # CVPR
    "https://openalex.org/S4363607764",  # ICCV
    "https://openalex.org/S4306419644",  # ICML
    "https://openalex.org/S4306419999",  # IJCAI
]

# Input file path and output file path
input_file = "/data/openalex/jsonl/works.txt"  # Replace with your input file path
output_file = "./openalex_data/ccf-ai-works.txt"

with open(output_file, "w", encoding="utf-8") as output_f:
    with open(input_file, "r", encoding="utf-8") as input_f:
        for line in input_f:
            try:
                data = json.loads(line.strip())
                # Check if the JSON data contains the primary_location and source fields.
                source_id = data["primary_location"]["source"]["id"]
                if source_id in valid_ids:
                    output_f.write(line)
            except Exception:
                # The current line is not valid JSON data or doesn't contain the primary_location and source fields.
                continue
```

```python
"""
Determined by "booktitle"
CSRankings Recommended Conference List
Please create output folder openalex_data at first.
"""

import json

# ID list
valid_ids = [
    "IJCAI",
    "NAACL",
    "EMNLP",
    "WWW",
    "ECCV",
    "ICCV",
    "ICML",
    "AAAI",
    "ICLR",
    "NeurIPS",
    "SIGIR",
    "KDD",
    "ACL",
    "CVPR",
]

# Input file path and output file path
input_file = "/data/openalex/jsonl/works.txt"  # Replace with your input file path
output_file = "./openalex_data/csr-ai-works.txt"

with open(output_file, "w", encoding="utf-8") as output_f:
    with open(input_file, "r", encoding="utf-8") as input_f:
        for line in input_f:
            try:
                data = json.loads(line.strip())
                # Check if the JSON data contains the booktitle fields.
                source_id = data["booktitle"]
                if source_id in valid_ids:
                    output_f.write(line)
            except Exception:
                # The current line is not valid JSON data or doesn't contain the booktitle fields.
                continue
```
